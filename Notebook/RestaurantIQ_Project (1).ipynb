## Define X and y

````python
# ============================================================
# DEFINE FEATURES AND TARGET
# ============================================================

target = "Aggregate rating"

numeric_features = [
    "Longitude",
    "Latitude",
    "Price range",
    "Log Votes",
    "Log Average Cost"
]

categorical_features = [
    "Country Code",
    "City",
    "Locality",
    "Cuisines",
    "Currency",
    "Has Table booking",
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu"
]

features = numeric_features + categorical_features

X = df_rating[features]
y = df_rating[target]

print(f"Number of features: {len(features)}")
print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
````

````python
# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training set:")
print(f"X_train: {X_train.shape}")
print(f"y_train: {y_train.shape}")

print("\nTesting set:")
print(f"X_test: {X_test.shape}")
print(f"y_test: {y_test.shape}")
````

### **Build preprocessing pipeline**



We will not manually encode categories before splitting.

Instead:


```
Training Data
     ↓
ColumnTransformer
     ├── Numerical → Imputation → Scaling
     │
     └── Categorical → Imputation → One-Hot Encoding
```



- The encoder is fitted only on the training data.

````python
# ============================================================
# PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Numerical preprocessing
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)

# Combine preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
````

````python
# ============================================================
# MACHINE LEARNING MODELS
# ============================================================

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)

models = {
    "Baseline": DummyRegressor(strategy="mean"),

    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=12,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=20,
        min_samples_leaf=2,
        n_jobs=-1,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=4,
        random_state=42
    )
}

print("Models defined:")
for name in models:
    print(" -", name)
````

````python
# ============================================================
# TRAIN AND EVALUATE REGRESSION MODELS
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

results = []
trained_models = {}

for name, model in models.items():

    print(f"Training {name}...")

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R²": r2
    })

    trained_models[name] = pipeline

    print(f"   MAE  : {mae:.4f}")
    print(f"   RMSE : {rmse:.4f}")
    print(f"   R²   : {r2:.4f}")
    print()

results_df = (
    pd.DataFrame(results)
    .sort_values("RMSE")
    .reset_index(drop=True)
)

print("MODEL COMPARISON")
print("=" * 70)

display(results_df)
````

````python
# ============================================================
# CLEAN FEATURE NAMES FOR INTERPRETATION
# ============================================================

def clean_feature_name(feature):
    """
    Convert transformed pipeline feature names
    into more readable names.
    """

    if feature.startswith("num__"):
        return feature.replace("num__", "")

    if feature.startswith("cat__"):
        feature = feature.replace("cat__", "")

        # Split categorical variable from category value
        parts = feature.split("_", 1)

        if len(parts) == 2:
            return f"{parts[0]} = {parts[1]}"

        return feature

    return feature


feature_importance_df["Readable Feature"] = (
    feature_importance_df["Feature"]
    .apply(clean_feature_name)
)

display(
    feature_importance_df[
        ["Readable Feature", "Importance"]
    ].head(30)
)
````

````python
# ============================================================
# 5-FOLD CROSS-VALIDATION - RANDOM FOREST
# ============================================================

from sklearn.model_selection import KFold, cross_validate

# Rebuild a fresh pipeline to ensure every fold
# performs preprocessing independently.
rf_cv_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                max_depth=20,
                min_samples_leaf=2,
                n_jobs=-1,
                random_state=42
            )
        )
    ]
)

# 5-fold cross-validation
kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    rf_cv_pipeline,
    X,
    y,
    cv=kf,
    scoring={
        "MAE": "neg_mean_absolute_error",
        "MSE": "neg_mean_squared_error",
        "R2": "r2"
    },
    n_jobs=-1,
    return_train_score=False
)

# Convert negative error scores to positive values
cv_mae = -cv_results["test_MAE"]
cv_mse = -cv_results["test_MSE"]
cv_rmse = np.sqrt(cv_mse)
cv_r2 = cv_results["test_R2"]

cv_summary = pd.DataFrame({
    "Metric": [
        "MAE",
        "RMSE",
        "R²"
    ],
    "Mean": [
        cv_mae.mean(),
        cv_rmse.mean(),
        cv_r2.mean()
    ],
    "Std": [
        cv_mae.std(),
        cv_rmse.std(),
        cv_r2.std()
    ]
})

cv_summary["Mean"] = cv_summary["Mean"].round(4)
cv_summary["Std"] = cv_summary["Std"].round(4)

print("5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 60)

display(cv_summary)
````

## Cuisine normalization

The cuisine field contains combinations such as:

```
North Indian, Chinese
Italian, Pizza, Bakery
```

For content-based recommendation, we'll normalize these strings.

````python
# ============================================================
# CUISINE TEXT NORMALIZATION
# ============================================================

def normalize_cuisines(cuisine):
    if pd.isna(cuisine):
        return ""

    cuisines = [
        c.strip().lower()
        for c in str(cuisine).split(",")
    ]

    return " ".join(cuisines)


df_recommend["Cuisine Text"] = (
    df_recommend["Cuisines"]
    .apply(normalize_cuisines)
)

display(
    df_recommend[
        ["Cuisines", "Cuisine Text"]
    ].head(10)
)
````

## Why we're doing this

Suppose the user likes:

> **North Indian**

A restaurant with:

> `North Indian, Chinese`

should be considered relevant.

A pure exact-match filter would miss that nuance.

Our content-based approach can recognize overlapping cuisine terms.


### Create the content representation

````python
# ============================================================
# TF-IDF REPRESENTATION OF CUISINES
# ============================================================

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    token_pattern=r"(?u)\b[\w&-]+\b"
)

cuisine_matrix = tfidf.fit_transform(
    df_recommend["Cuisine Text"]
)

print("Cuisine TF-IDF matrix created.")
print(f"Matrix shape: {cuisine_matrix.shape}")
````

### Create recommendation function

````python
# ============================================================
# CONTENT-BASED RESTAURANT RECOMMENDER
# ============================================================

def recommend_restaurants(
    cuisine=None,
    city=None,
    price_range=None,
    min_rating=0,
    online_delivery=None,
    table_booking=None,
    top_n=10
):
    """
    Recommend restaurants based on user preferences.
    """

    recommendations = df_recommend.copy()

    # --------------------------------------------------------
    # Hard filters
    # --------------------------------------------------------

    if city is not None:
        recommendations = recommendations[
            recommendations["City"].str.lower()
            == city.lower()
        ]

    if price_range is not None:
        recommendations = recommendations[
            recommendations["Price range"]
            <= price_range
        ]

    if min_rating > 0:
        recommendations = recommendations[
            recommendations["Aggregate rating"]
            >= min_rating
        ]

    if online_delivery is not None:
        recommendations = recommendations[
            recommendations["Has Online delivery"]
            .str.lower()
            == online_delivery.lower()
        ]

    if table_booking is not None:
        recommendations = recommendations[
            recommendations["Has Table booking"]
            .str.lower()
            == table_booking.lower()
        ]

    # --------------------------------------------------------
    # Cuisine relevance
    # --------------------------------------------------------

    if cuisine is not None:

        cuisine_terms = [
            term.strip().lower()
            for term in cuisine.split(",")
        ]

        def cuisine_match(cuisine_text):
            return sum(
                term in cuisine_text
                for term in cuisine_terms
            )

        recommendations["Cuisine Match"] = (
            recommendations["Cuisine Text"]
            .apply(cuisine_match)
        )

    else:
        recommendations["Cuisine Match"] = 0

    # --------------------------------------------------------
    # Recommendation score
    # --------------------------------------------------------

    recommendations["Recommendation Score"] = (
        recommendations["Aggregate rating"] * 0.60
        +
        np.log1p(
            recommendations["Votes"]
        ) * 0.10
        +
        recommendations["Cuisine Match"] * 0.30
    )

    # Sort
    recommendations = (
        recommendations
        .sort_values(
            "Recommendation Score",
            ascending=False
        )
        .head(top_n)
    )

    # Output columns
    output_columns = [
        "Restaurant Name",
        "City",
        "Locality",
        "Cuisines",
        "Price range",
        "Average Cost for two",
        "Aggregate rating",
        "Votes",
        "Has Online delivery",
        "Has Table booking",
        "Recommendation Score"
    ]

    return recommendations[output_columns]
````

## Problem

Our current formula is:

```text
Recommendation Score =
    60% Rating
  + 10% Log Votes
  + 30% Cuisine Match
```

This is not really a proper **content-based filtering** system.

We need to calculate **similarity between the user's preferred cuisine and each restaurant's cuisine profile**.

# Step 1 — Build proper cosine-similarity recommendation

We already created:

```text
cuisine_matrix
```

which has:

```text
(9551, 152)
```

Now we'll use it.


````python
# ============================================================
# TASK 2 - CONTENT-BASED COSINE SIMILARITY RECOMMENDER
# ============================================================

from sklearn.metrics.pairwise import cosine_similarity


def recommend_restaurants_v2(
    cuisine=None,
    city=None,
    price_range=None,
    min_rating=0,
    online_delivery=None,
    table_booking=None,
    top_n=10
):
    """
    Content-based restaurant recommendation system.

    Uses:
    - TF-IDF cuisine representation
    - Cosine similarity
    - Rating quality
    - Popularity
    - User constraints
    """

    # --------------------------------------------------------
    # Create user cuisine profile
    # --------------------------------------------------------

    if cuisine is not None:
        user_cuisine = normalize_cuisines(cuisine)

        user_vector = tfidf.transform(
            [user_cuisine]
        )

        similarity_scores = cosine_similarity(
            user_vector,
            cuisine_matrix
        ).flatten()

    else:
        similarity_scores = np.zeros(
            len(df_recommend)
        )

    # --------------------------------------------------------
    # Work on complete recommendation dataframe
    # --------------------------------------------------------

    recommendations = df_recommend.copy()

    recommendations["Cuisine Similarity"] = (
        similarity_scores
    )

    # --------------------------------------------------------
    # Hard filters
    # --------------------------------------------------------

    if city is not None:
        recommendations = recommendations[
            recommendations["City"].str.lower()
            == city.lower()
        ]

    if price_range is not None:
        recommendations = recommendations[
            recommendations["Price range"]
            <= price_range
        ]

    if min_rating > 0:
        recommendations = recommendations[
            recommendations["Aggregate rating"]
            >= min_rating
        ]

    if online_delivery is not None:
        recommendations = recommendations[
            recommendations["Has Online delivery"]
            .str.lower()
            == online_delivery.lower()
        ]

    if table_booking is not None:
        recommendations = recommendations[
            recommendations["Has Table booking"]
            .str.lower()
            == table_booking.lower()
        ]

    # --------------------------------------------------------
    # Normalize rating
    # --------------------------------------------------------

    recommendations["Rating Normalized"] = (
        recommendations["Aggregate rating"] / 5.0
    )

    # --------------------------------------------------------
    # Normalize popularity
    # --------------------------------------------------------

    max_log_votes = (
        df_recommend["Log Votes"].max()
    )

    recommendations["Popularity Normalized"] = (
        recommendations["Log Votes"]
        / max_log_votes
    )

    # --------------------------------------------------------
    # Final recommendation score
    #
    # Cuisine similarity is dominant.
    # --------------------------------------------------------

    recommendations["Recommendation Score"] = (
        recommendations["Cuisine Similarity"] * 0.60
        +
        recommendations["Rating Normalized"] * 0.30
        +
        recommendations["Popularity Normalized"] * 0.10
    )

    # --------------------------------------------------------
    # Sort
    # --------------------------------------------------------

    recommendations = (
        recommendations
        .sort_values(
            [
                "Recommendation Score",
                "Aggregate rating",
                "Votes"
            ],
            ascending=False
        )
        .head(top_n)
    )

    # --------------------------------------------------------
    # Output
    # --------------------------------------------------------

    output_columns = [
        "Restaurant Name",
        "City",
        "Locality",
        "Cuisines",
        "Price range",
        "Average Cost for two",
        "Aggregate rating",
        "Votes",
        "Has Online delivery",
        "Has Table booking",
        "Cuisine Similarity",
        "Recommendation Score"
    ]

    return recommendations[output_columns]
````

# Add a recommendation explanation

This will make the project considerably better.

we can explain **why** it was recommended.

For example:

```text id="j4u9u2"
Recommended because:
✓ Cuisine matches user preference
✓ Rating ≥ 4.0
✓ Within preferred price range
✓ Online delivery available
```


````python
# ============================================================
# RECOMMENDATION EXPLANATION
# ============================================================

def add_recommendation_explanation(
    recommendations,
    requested_cuisine=None,
    requested_price=None,
    min_rating=0,
    online_delivery=None
):

    explanations = []

    for _, row in recommendations.iterrows():

        reasons = []

        # Cuisine
        if requested_cuisine is not None:
            if row["Cuisine Similarity"] > 0:
                reasons.append("Cuisine preference matched")

        # Rating
        if row["Aggregate rating"] >= min_rating:
            reasons.append(
                f"Rating ≥ {min_rating}"
            )

        # Price
        if requested_price is not None:
            if row["Price range"] <= requested_price:
                reasons.append(
                    "Within preferred price range"
                )

        # Delivery
        if online_delivery is not None:
            if (
                row["Has Online delivery"].lower()
                == online_delivery.lower()
            ):
                reasons.append(
                    "Online delivery available"
                )

        explanations.append(
            " • ".join(reasons)
        )

    recommendations = recommendations.copy()

    recommendations["Why Recommended"] = (
        explanations
    )

    return recommendations
````

# Recommendation quality evaluation

There is an important conceptual issue here.

A recommendation system needs some notion of **relevance**.

We don't have explicit user ratings/preferences in this dataset, so we cannot honestly claim:

> "The recommender has 95% accuracy."

That would be fabricated.

Instead, we can perform **rule-based offline evaluation**.

For each recommendation, check:

### Criterion 1 — Cuisine relevance

Does the restaurant contain the requested cuisine?

### Criterion 2 — Rating requirement

Does:

```text id="q5w8p2"
Aggregate rating ≥ minimum rating
```

### Criterion 3 — Price constraint

Does:

```text id="a1k7mz"
Price range ≤ user's maximum
```

### Criterion 4 — Delivery requirement

Does the restaurant provide delivery when requested?


### Recommendation evaluation

````python
# ============================================================
# TASK 2 - FINAL RECOMMENDATION EVALUATION
# ============================================================

def evaluate_recommendations_final(
    recommendations,
    requested_cuisine=None,
    preferred_price_range=None,
    min_rating=0,
    online_delivery=None
):

    evaluation = recommendations.copy()

    # --------------------------------------------------------
    # 1. Cuisine relevance
    # --------------------------------------------------------

    if requested_cuisine is not None:

        requested_terms = [
            x.strip().lower()
            for x in requested_cuisine.split(",")
        ]

        evaluation["Cuisine Relevant"] = (
            evaluation["Cuisines"]
            .fillna("")
            .str.lower()
            .apply(
                lambda x: any(
                    term in x
                    for term in requested_terms
                )
            )
        )

    else:

        evaluation["Cuisine Relevant"] = True

    # --------------------------------------------------------
    # 2. Minimum rating requirement
    # --------------------------------------------------------

    evaluation["Rating Requirement"] = (
        evaluation["Aggregate rating"] >= min_rating
    )

    # --------------------------------------------------------
    # 3. Price compatibility
    #
    # SOFT PREFERENCE - NOT A HARD REQUIREMENT
    # --------------------------------------------------------

    if preferred_price_range is not None:

        evaluation["Price Compatibility"] = (
            1 -
            (
                abs(
                    evaluation["Price range"]
                    - preferred_price_range
                ) / 3
            )
        ).clip(lower=0)

    else:

        evaluation["Price Compatibility"] = 1.0

    # --------------------------------------------------------
    # 4. Online delivery requirement
    # --------------------------------------------------------

    if online_delivery is not None:

        evaluation["Delivery Requirement"] = (
            evaluation["Has Online delivery"]
            .str.lower()
            == online_delivery.lower()
        )

    else:

        evaluation["Delivery Requirement"] = True

    # --------------------------------------------------------
    # 5. HARD-CONSTRAINT SATISFACTION
    #
    # Price is deliberately excluded because it is
    # a soft preference.
    # --------------------------------------------------------

    evaluation["Meets Hard Criteria"] = (
        evaluation["Cuisine Relevant"]
        &
        evaluation["Rating Requirement"]
        &
        evaluation["Delivery Requirement"]
    )

    return evaluation
````

## Improved final scoring

````python
# ============================================================
# TASK 2 - FINAL RECOMMENDATION SCORING
# ============================================================

def recommend_restaurants_final(
    cuisine=None,
    city=None,
    price_range=None,
    min_rating=0,
    online_delivery=None,
    table_booking=None,
    top_n=10
):

    # --------------------------------------------------------
    # Cuisine similarity
    # --------------------------------------------------------

    if cuisine is not None:

        user_cuisine = normalize_cuisines(cuisine)

        user_vector = tfidf.transform(
            [user_cuisine]
        )

        similarity_scores = cosine_similarity(
            user_vector,
            cuisine_matrix
        ).flatten()

    else:
        similarity_scores = np.zeros(
            len(df_recommend)
        )

    recommendations = df_recommend.copy()

    recommendations["Cuisine Similarity"] = (
        similarity_scores
    )

    # --------------------------------------------------------
    # City filter
    # --------------------------------------------------------

    if city is not None:
        recommendations = recommendations[
            recommendations["City"].str.lower()
            == city.lower()
        ]

    # --------------------------------------------------------
    # Minimum rating
    # --------------------------------------------------------

    if min_rating > 0:
        recommendations = recommendations[
            recommendations["Aggregate rating"]
            >= min_rating
        ]

    # --------------------------------------------------------
    # Delivery filter
    # --------------------------------------------------------

    if online_delivery is not None:
        recommendations = recommendations[
            recommendations["Has Online delivery"]
            .str.lower()
            == online_delivery.lower()
        ]

    # --------------------------------------------------------
    # Table booking filter
    # --------------------------------------------------------

    if table_booking is not None:
        recommendations = recommendations[
            recommendations["Has Table booking"]
            .str.lower()
            == table_booking.lower()
        ]

    # --------------------------------------------------------
    # Rating normalization
    # --------------------------------------------------------

    recommendations["Rating Quality"] = (
        recommendations["Aggregate rating"] / 5
    )

    # --------------------------------------------------------
    # Popularity normalization
    # --------------------------------------------------------

    max_votes = (
        df_recommend["Log Votes"].max()
    )

    recommendations["Popularity"] = (
        recommendations["Log Votes"]
        / max_votes
    )

    # --------------------------------------------------------
    # Price compatibility
    # --------------------------------------------------------

    if price_range is not None:

        recommendations["Price Compatibility"] = (
            1 -
            (
                abs(
                    recommendations["Price range"]
                    - price_range
                ) / 3
            )
        ).clip(lower=0)

    else:

        recommendations["Price Compatibility"] = 1

    # --------------------------------------------------------
    # Service preference
    # --------------------------------------------------------

    recommendations["Service Score"] = 1.0

    if online_delivery is not None:

        recommendations["Service Score"] = (
            recommendations["Has Online delivery"]
            .str.lower()
            == online_delivery.lower()
        ).astype(float)

    if table_booking is not None:

        table_score = (
            recommendations["Has Table booking"]
            .str.lower()
            == table_booking.lower()
        ).astype(float)

        recommendations["Service Score"] = (
            recommendations["Service Score"]
            + table_score
        ) / 2

    # --------------------------------------------------------
    # Final weighted score
    # --------------------------------------------------------

    recommendations["Recommendation Score"] = (

        recommendations["Cuisine Similarity"] * 0.50

        +

        recommendations["Rating Quality"] * 0.25

        +

        recommendations["Popularity"] * 0.10

        +

        recommendations["Price Compatibility"] * 0.10

        +

        recommendations["Service Score"] * 0.05
    )

    # --------------------------------------------------------
    # Sort
    # --------------------------------------------------------

    recommendations = (
        recommendations
        .sort_values(
            [
                "Recommendation Score",
                "Cuisine Similarity",
                "Aggregate rating"
            ],
            ascending=False
        )
        .head(top_n)
    )

    # --------------------------------------------------------
    # Output
    # --------------------------------------------------------

    output_columns = [
        "Restaurant Name",
        "City",
        "Locality",
        "Cuisines",
        "Price range",
        "Average Cost for two",
        "Aggregate rating",
        "Votes",
        "Has Online delivery",
        "Has Table booking",
        "Cuisine Similarity",
        "Price Compatibility",
        "Recommendation Score"
    ]

    return recommendations[output_columns]
````

````python
# ============================================================
# RARE CUISINE ANALYSIS
# ============================================================

rare_thresholds = [1, 2, 5, 10, 20, 50, 100]

print("=" * 70)
print("RARE CUISINE ANALYSIS")
print("=" * 70)

for threshold in rare_thresholds:

    number_labels = (
        (cuisine_frequency >= threshold)
        .sum()
    )

    restaurants_covered = (
        cuisine_frequency[
            cuisine_frequency >= threshold
        ].sum()
    )

    print(
        f"Minimum frequency >= {threshold:3d}: "
        f"{number_labels:3d} cuisines | "
        f"{restaurants_covered:5d} restaurant-label assignments"
    )
````

### Keep learnable cuisine classes

````python
# ============================================================
# TASK 3 — STEP 5
# FILTER RARE CUISINE LABELS
# ============================================================

MIN_CUISINE_FREQUENCY = 20

# Keep cuisine labels appearing at least 20 times
selected_cuisines = (
    cuisine_frequency[
        cuisine_frequency >= MIN_CUISINE_FREQUENCY
    ]
    .index
    .tolist()
)

print("=" * 70)
print("CUISINE LABEL FILTERING")
print("=" * 70)

print(f"Original cuisine labels : {len(cuisine_frequency)}")
print(f"Selected cuisine labels : {len(selected_cuisines)}")
print(f"Minimum frequency       : {MIN_CUISINE_FREQUENCY}")

print("\nSelected cuisines:")
print(selected_cuisines)
````

Create the final multi-label target

We need to be careful here.

````python
# ============================================================
# TASK 3 — STEP 6
# CREATE FILTERED MULTI-LABEL TARGET
# ============================================================

filtered_cuisine_lists = []

for cuisine_list in cuisine_lists:

    filtered_labels = [
        cuisine
        for cuisine in cuisine_list
        if cuisine in selected_cuisines
    ]

    filtered_cuisine_lists.append(filtered_labels)

# Remove restaurants with no remaining learnable cuisine label
valid_mask = np.array([
    len(labels) > 0
    for labels in filtered_cuisine_lists
])

X_task3_base = df_task3.loc[valid_mask].copy()

y_cuisine_lists = [
    labels
    for labels, valid in zip(filtered_cuisine_lists, valid_mask)
    if valid
]

# Encode target
mlb_task3 = MultiLabelBinarizer()

y_task3 = mlb_task3.fit_transform(y_cuisine_lists)

print("=" * 70)
print("FINAL TASK 3 TARGET")
print("=" * 70)

print("Original restaurants :", len(df_task3))
print("Final restaurants    :", len(X_task3_base))
print("Cuisine labels       :", y_task3.shape[1])
print("Target matrix shape  :", y_task3.shape)

print("\nLabels:")
print(list(mlb_task3.classes_))
````

# Build preprocessing pipeline

We'll use:

* **Median imputation + scaling** for numerical variables
* **Most-frequent imputation + One-Hot Encoding** for categorical variables

For the classifier, we'll start with **One-vs-Rest Logistic Regression**.

Why?

Because this is a strong baseline for high-dimensional sparse one-hot encoded data and gives us a clean, academically defensible multi-label classifier.


````python
# ============================================================
# TASK 3 — STEP 11
# PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_transformer_task3 = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer_task3 = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                min_frequency=2
            )
        )
    ]
)

preprocessor_task3 = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer_task3,
            numeric_features_task3
        ),
        (
            "cat",
            categorical_transformer_task3,
            categorical_features_task3
        )
    ]
)

print("Preprocessing pipeline created successfully.")
````

## Train the first classifier

Now we'll use:

- One-vs-Rest Logistic Regression

````python
# ============================================================
# TASK 3 — STEP 12
# MULTI-LABEL LOGISTIC REGRESSION
# ============================================================

from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression

classifier_task3 = OneVsRestClassifier(
    LogisticRegression(
        max_iter=2000,
        C=1.0,
        class_weight="balanced",
        solver="liblinear"
    )
)

model_task3_lr = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor_task3
        ),
        (
            "classifier",
            classifier_task3
        )
    ]
)

print("Training One-vs-Rest Logistic Regression...")

model_task3_lr.fit(
    X_train_task3,
    y_train_task3
)

print("Training completed.")
````

### Generate predictions

````python
# ============================================================
# TASK 3 — STEP 13
# PREDICTIONS
# ============================================================

y_pred_task3_lr = model_task3_lr.predict(
    X_test_task3
)

print("=" * 70)
print("PREDICTION RESULTS")
print("=" * 70)

print("Actual target shape    :", y_test_task3.shape)
print("Predicted target shape :", y_pred_task3_lr.shape)

print("\nTotal test restaurants:")
print(len(y_test_task3))
````

### Evaluate the classifier

For a multi-label problem, accuracy alone is insufficient.

We'll calculate:
```
Exact Match Accuracy
Hamming Loss
Micro Precision
Micro Recall
Micro F1
Macro Precision
Macro Recall
Macro F1
Weighted F1
```

````python
# ============================================================
# TASK 3 — STEP 14
# CLASSIFICATION METRICS
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    hamming_loss,
    precision_score,
    recall_score,
    f1_score
)

exact_match_accuracy = accuracy_score(
    y_test_task3,
    y_pred_task3_lr
)

hamming = hamming_loss(
    y_test_task3,
    y_pred_task3_lr
)

micro_precision = precision_score(
    y_test_task3,
    y_pred_task3_lr,
    average="micro",
    zero_division=0
)

micro_recall = recall_score(
    y_test_task3,
    y_pred_task3_lr,
    average="micro",
    zero_division=0
)

micro_f1 = f1_score(
    y_test_task3,
    y_pred_task3_lr,
    average="micro",
    zero_division=0
)

macro_precision = precision_score(
    y_test_task3,
    y_pred_task3_lr,
    average="macro",
    zero_division=0
)

macro_recall = recall_score(
    y_test_task3,
    y_pred_task3_lr,
    average="macro",
    zero_division=0
)

macro_f1 = f1_score(
    y_test_task3,
    y_pred_task3_lr,
    average="macro",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test_task3,
    y_pred_task3_lr,
    average="weighted",
    zero_division=0
)

task3_results_lr = pd.DataFrame({
    "Metric": [
        "Exact Match Accuracy",
        "Hamming Loss",
        "Micro Precision",
        "Micro Recall",
        "Micro F1",
        "Macro Precision",
        "Macro Recall",
        "Macro F1",
        "Weighted F1"
    ],
    "Value": [
        exact_match_accuracy,
        hamming,
        micro_precision,
        micro_recall,
        micro_f1,
        macro_precision,
        macro_recall,
        macro_f1,
        weighted_f1
    ]
})

display(task3_results_lr)
````

## Why we're using these metrics


### Exact Match Accuracy

A prediction is correct only if **every cuisine label** is correctly predicted for that restaurant.

This is strict.

### Hamming Loss

Measures the fraction of incorrect individual labels.

**Lower is better.**

### Micro F1

Weights all cuisine-label decisions equally and is useful when the dataset is imbalanced.

### Macro F1

Calculates F1 independently for each cuisine and then averages them.

This is particularly important here because:

> North Indian has 3,960 examples, while some cuisines have only 20–50.

Macro F1 therefore helps expose whether the model performs well only on dominant cuisines.

# ⚠️ LIMITATIONS


### 1. Rating leakage / temporal interpretation

`Votes` is extremely predictive of rating, but the relationship may reflect accumulated user engagement and therefore should not automatically be interpreted as a causal determinant of rating.

### 2. Class imbalance

The cuisine classification task contains substantial imbalance across the 57 cuisine categories. Rare cuisines have insufficient observations for reliable classification.

### 3. Recommendation personalization

The recommendation system uses explicitly provided preferences rather than historical user behavior. Therefore, it is a **content-based recommender**, not a collaborative filtering system.

### 4. Geographic concentration

The restaurant dataset is heavily concentrated in particular cities, especially New Delhi, Gurgaon, and Noida. Consequently, geographic patterns should not necessarily be generalized to the global restaurant market.

### 5. Dataset vintage

Restaurant ratings, prices, delivery availability, and restaurant operations can change over time. Therefore, the results should be interpreted as reflecting the dataset's underlying observation period rather than current restaurant conditions.

### 6. Rating subjectivity

Restaurant ratings are subjective user-generated measurements and may be affected by customer preferences, review behavior, restaurant popularity, and sampling effects.
